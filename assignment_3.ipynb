```
# Install packages
import threading, subprocess, time, ollama
```

```
def run_ollama_serve():
    """
    Lets Ollama run service in the background parallel to other code blocks.
    """
    subprocess.Popen(["ollama", "serve"])

thread = threading.Thread(target=run_ollama_serve)
thread.start()
time.sleep(5)
```
